$F = \frac{9}{5}C + 32$

```python
celcius = float(input('Digite a temperatura em celcius: '))
fahrenheit = (celcius * 9 / 5) + 32

print(f"Temperatura em Fahrenheit: {fahrenheit:.2f}°F")
```

#### 10. Tendo como dados de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, usando a seguinte fórmula:
$P = 72,7h - 58$

#### 11. Tendo como dado de entrada a altura (h) de uma pessoa, construa um algoritmo que calcule seu peso ideal, utilizando as seguintes fórmulas:
##### a. Para homens: $P = 72,7h - 58$
##### b. Para mulheres: $P = 62,1h - 44,7$

```python
h = float(input('Digite sua altura em metros: '))

homens = (72.7 * h) - 58
mulheres = (62.1 * h) - 44.7

print(f'Peso ideal para homens {homens:.2f} kg')
print(f'Peso ideal para mulheres {mulheres:.2f} kg')
```

#### 12. Faça um Programa que pergunte quanto você ganha por hora e o número de horas trabalhadas no mês.

```python
ganho_por_hora_trabalhada = float(input('Digite quanto ganha por hora trabalhada: '))

dias_do_mes_trabalho = 22
horas_dia = 8
horas_trabalhadas_mes = horas_dia * dias_do_mes_trabalho
```

#####  Calcule o salário bruto (horas * salario por hora)

```python
salario_bruto = horas_trabalhadas_mes * ganho_por_hora_trabalhada
print(f'Meu salario bruto é R$ {salario_bruto} reais')
```

##### Calcule o desconto do IR (11% do salário bruto)

```python
desconto_ir = salario_bruto * 0.11
print(f'Meu desconto do IR é de R$ {desconto_ir:.2f} reais')
```

##### Calcule o desconto do INSS (8% do salário bruto)

```python
desconto_inss = salario_bruto * 0.08
print(f'Meu desconto do INSS é de R$ {desconto_inss:.2f} reais')
```

##### Calcule o desconto do sindicato (5% do salário bruto)

```python
desconto_sindicato = salario_bruto * 0.05
print(f'Meu desconto do sindicato é de R$ {desconto_sindicato:.2f} reais')
```

##### Calcule o salário líquido (salário bruto - descontos)

```python
salario_liquido = salario_bruto - desconto_inss - desconto_ir - desconto_sindicato
soma_desconto = desconto_ir + desconto_inss + desconto_sindicato
print(f'Meu salario bruto é de R$ {salario_bruto:.2f} reais')
print(f'Meu desconto INSS é de R$ {desconto_inss:.2f} reais')
print(f'Meu desconto IR é de R$ {desconto_ir:.2f} reais')
print(f'Meu desconto sindicato é de R$ {desconto_sindicato:.2f} reais')
print(f'Meus descontos somados R$ {soma_desconto:.2f} reais')
print(f'Meu salario liquido é de R$ {salario_liquido:.2f} reais')
```
